```python
import os
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from sklearn.model_selection import (
    train_test_split,
    KFold,
    cross_val_score
)

from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder
)

from sklearn.compose import ColumnTransformer

from sklearn.pipeline import Pipeline

from sklearn.impute import SimpleImputer

from sklearn.linear_model import LinearRegression

from sklearn.tree import DecisionTreeRegressor

from sklearn.ensemble import (
    RandomForestRegressor,
    GradientBoostingRegressor
)

from sklearn.svm import SVR

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)


warnings.filterwarnings("ignore")

sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (12, 7)

plt.rcParams["font.size"] = 11

np.random.seed(42)


# ================================================================
# 2. PROJECT DIRECTORY
# ================================================================

BASE_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based State-Wise Widowed and Divorced Women Population Analysis"
)

DATASET_PATH = os.path.join(
    BASE_DIR,
    "RS_Session_259_AU_3352_A_and_B.csv"
)

os.makedirs(BASE_DIR, exist_ok=True)


def save_and_show(filename):

    """
    Save the current graph and display it.
    """

    output_path = os.path.join(
        BASE_DIR,
        filename
    )

    plt.tight_layout()

    plt.savefig(
        output_path,
        dpi=300,
        bbox_inches="tight"
    )

    print(f"\n[SAVED] {output_path}")

    plt.show()

    plt.close()


print("=" * 70)

print("AI-BASED WOMEN POPULATION ANALYSIS")

print("=" * 70)


# ================================================================
# 3. LOAD DATASET
# ================================================================

print("\nLoading Dataset...")

try:

    df = pd.read_csv(
        DATASET_PATH,
        encoding="utf-8-sig"
    )

except UnicodeDecodeError:

    df = pd.read_csv(
        DATASET_PATH,
        encoding="latin1"
    )


print("\nDataset Loaded Successfully!")

print("\nDataset Shape:")

print(df.shape)

print("\nDataset Columns:")

print(df.columns.tolist())

print("\nFirst Five Rows:")

print(df.head())


# ================================================================
# 4. DATA PREPROCESSING
# ================================================================

print("\nPerforming Data Preprocessing...")


df.columns = df.columns.str.strip()


STATE_COLUMN = "State/ UT"

WIDOWED_COLUMN = "Female of all Ages - Widowed"

DIVORCED_COLUMN = "Female of all Ages - Divorced"


required_columns = [
    STATE_COLUMN,
    WIDOWED_COLUMN,
    DIVORCED_COLUMN
]


missing_columns = [
    col for col in required_columns
    if col not in df.columns
]


if missing_columns:

    raise ValueError(
        f"Missing columns: {missing_columns}\n"
        f"Available columns: {df.columns.tolist()}"
    )


# Remove duplicate records

df = df.drop_duplicates()


# Clean state names

df[STATE_COLUMN] = (
    df[STATE_COLUMN]
    .astype("string")
    .str.strip()
)


# Convert population columns to numeric

for column in [
    WIDOWED_COLUMN,
    DIVORCED_COLUMN
]:

    df[column] = pd.to_numeric(

        df[column]
        .astype(str)
        .str.replace(",", "", regex=False),

        errors="coerce"
    )


# Remove rows with missing values

df = df.dropna(
    subset=[
        STATE_COLUMN,
        DIVORCED_COLUMN
    ]
)


# Remove negative population values

df = df[
    df[DIVORCED_COLUMN] >= 0
]


df = df.reset_index(drop=True)


if len(df) < 10:

    raise ValueError(
        "Dataset contains too few valid records."
    )


print("\nCleaned Dataset Shape:")

print(df.shape)


print("\nMissing Values:")

print(df.isnull().sum())


print("\nStatistical Summary:")

print(df.describe())


# ================================================================
# 5. DEFINE FEATURES AND TARGET
# ================================================================

print("\nPreparing Features and Target...")


X = df[
    [
        STATE_COLUMN,
        WIDOWED_COLUMN
    ]
]


y = df[DIVORCED_COLUMN].astype(float)


# ================================================================
# 6. TRAIN TEST SPLIT
# ================================================================

X_train, X_test, y_train, y_test = train_test_split(

    X,

    y,

    test_size=0.25,

    random_state=42
)


print("\nTraining Records:", len(X_train))

print("Testing Records:", len(X_test))


# ================================================================
# 7. PREPROCESSING PIPELINE
# ================================================================

numeric_features = [
    WIDOWED_COLUMN
]

categorical_features = [
    STATE_COLUMN
]


numeric_transformer = Pipeline(

    steps=[

        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),

        (
            "scaler",
            StandardScaler()
        )

    ]
)


categorical_transformer = Pipeline(

    steps=[

        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),

        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )

    ]
)


def create_preprocessor():

    return ColumnTransformer(

        transformers=[

            (
                "numeric",
                numeric_transformer,
                numeric_features
            ),

            (
                "categorical",
                categorical_transformer,
                categorical_features
            )

        ]
    )


# ================================================================
# 8. DEFINE MACHINE LEARNING MODELS
# ================================================================

models = {

    "Linear Regression":

        LinearRegression(),

    "Decision Tree":

        DecisionTreeRegressor(
            max_depth=4,
            random_state=42
        ),

    "Random Forest":

        RandomForestRegressor(
            n_estimators=100,
            max_depth=5,
            random_state=42
        ),

    "Gradient Boosting":

        GradientBoostingRegressor(
            n_estimators=100,
            learning_rate=0.05,
            max_depth=3,
            random_state=42
        ),

    "Support Vector Regression":

        SVR(
            kernel="rbf",
            C=100,
            epsilon=0.1
        )

}


# ================================================================
# 9. TRAIN AND EVALUATE MODELS
# ================================================================

print("\nTraining Machine Learning Models...")


results = []

trained_models = {}


for model_name, model in models.items():

    print("\n" + "-" * 60)

    print("Training:", model_name)

    print("-" * 60)


    pipeline = Pipeline(

        steps=[

            (
                "preprocessor",
                create_preprocessor()
            ),

            (
                "regressor",
                model
            )

        ]
    )


    # Train model

    pipeline.fit(
        X_train,
        y_train
    )


    # Predict test values

    y_pred = pipeline.predict(
        X_test
    )


    # Population cannot be negative

    y_pred = np.maximum(
        y_pred,
        0
    )


    # Evaluation metrics

    mae = mean_absolute_error(
        y_test,
        y_pred
    )


    mse = mean_squared_error(
        y_test,
        y_pred
    )


    rmse = np.sqrt(mse)


    r2 = r2_score(
        y_test,
        y_pred
    )


    # Cross validation

    cv = KFold(
        n_splits=5,
        shuffle=True,
        random_state=42
    )


    cv_scores = cross_val_score(

        pipeline,

        X_train,

        y_train,

        cv=cv,

        scoring="neg_mean_absolute_error"
    )


    cv_mae = -np.mean(
        cv_scores
    )


    results.append({

        "Model": model_name,

        "MAE": float(mae),

        "MSE": float(mse),

        "RMSE": float(rmse),

        "R2": float(r2),

        "CV_MAE": float(cv_mae)

    })


    trained_models[model_name] = pipeline


    print("MAE:", round(mae, 4))

    print("MSE:", round(mse, 4))

    print("RMSE:", round(rmse, 4))

    print("R2:", round(r2, 4))

    print("Cross Validation MAE:", round(cv_mae, 4))


# ================================================================
# 10. CREATE RESULTS DATAFRAME
# ================================================================

results_df = pd.DataFrame(
    results
)


results_df = results_df.sort_values(
    by="CV_MAE",
    ascending=True
)


print("\nModel Comparison Results:")

print(
    results_df.to_string(
        index=False
    )
)


# ================================================================
# 11. SELECT BEST MODEL
# ================================================================

best_model_name = results_df.iloc[0]["Model"]


best_model = trained_models[
    best_model_name
]


print("\n" + "=" * 70)

print("BEST MODEL:", best_model_name)

print("=" * 70)


# ================================================================
# 12. ACCURACY GRAPH
# ================================================================

print("\nGenerating Accuracy Graph...")


plt.figure(
    figsize=(12, 7)
)


# R2 score is used instead of classification accuracy

accuracy_data = results_df.sort_values(
    by="R2",
    ascending=False
)


bars = plt.bar(

    accuracy_data["Model"],

    accuracy_data["R2"],

    color=sns.color_palette(
        "viridis",
        len(accuracy_data)
    ),

    edgecolor="black"
)


plt.title(

    "Machine Learning Model Performance Comparison (R²)",

    fontsize=16,

    fontweight="bold"
)


plt.xlabel(
    "Machine Learning Algorithms",
    fontsize=12
)


plt.ylabel(
    "R² Score",
    fontsize=12
)


plt.xticks(
    rotation=25,
    ha="right"
)


plt.axhline(
    y=0,
    color="black",
    linestyle="--",
    linewidth=1
)


for bar in bars:

    height = bar.get_height()

    offset = max(
        abs(height) * 0.03,
        0.02
    )

    plt.text(

        bar.get_x() + bar.get_width() / 2,

        height + (
            offset if height >= 0 else -offset
        ),

        f"{height:.3f}",

        ha="center",

        va=(
            "bottom"
            if height >= 0
            else "top"
        ),

        fontsize=10
    )


plt.grid(
    axis="y",
    alpha=0.3
)


save_and_show(
    "accuracy_graph.png"
)


# ================================================================
# 13. CORRELATION HEATMAP
# ================================================================

print("\nGenerating Correlation Heatmap...")


numeric_df = df[
    [
        WIDOWED_COLUMN,
        DIVORCED_COLUMN
    ]
]


correlation_matrix = numeric_df.corr()


plt.figure(
    figsize=(10, 7)
)


sns.heatmap(

    correlation_matrix,

    annot=True,

    cmap="coolwarm",

    fmt=".3f",

    linewidths=1,

    linecolor="white",

    square=True,

    vmin=-1,

    vmax=1,

    cbar_kws={
        "label": "Correlation Coefficient"
    }
)


plt.title(

    "Correlation Heatmap of Widowed and Divorced Women Population",

    fontsize=15,

    fontweight="bold"
)


plt.xticks(
    rotation=20,
    ha="right"
)


plt.yticks(
    rotation=0
)


save_and_show(
    "heatmap.png"
)


# ================================================================
# 14. MODEL COMPARISON GRAPH
# ================================================================

print("\nGenerating Model Comparison Graph...")


comparison_df = results_df.sort_values(
    by="RMSE",
    ascending=True
)


x = np.arange(
    len(comparison_df)
)


width = 0.35


plt.figure(
    figsize=(13, 7)
)


bars1 = plt.bar(

    x - width / 2,

    comparison_df["MAE"],

    width,

    label="MAE",

    color="steelblue",

    edgecolor="black"
)


bars2 = plt.bar(

    x + width / 2,

    comparison_df["RMSE"],

    width,

    label="RMSE",

    color="coral",

    edgecolor="black"
)


plt.xlabel(
    "Machine Learning Algorithms",
    fontsize=12
)


plt.ylabel(
    "Prediction Error (Women)",
    fontsize=12
)


plt.title(

    "Comparison of Machine Learning Models Using MAE and RMSE",

    fontsize=16,

    fontweight="bold"
)


plt.xticks(

    x,

    comparison_df["Model"],

    rotation=25,

    ha="right"
)


plt.legend()


plt.grid(
    axis="y",
    alpha=0.3
)


save_and_show(
    "comparison_graph.png"
)


# ================================================================
# 15. GENERATE RESULT CSV
# ================================================================

print("\nGenerating Result CSV...")


best_predictions = best_model.predict(
    X_test
)


best_predictions = np.maximum(
    best_predictions,
    0
)


result_df = pd.DataFrame({

    "State/UT":

        X_test[STATE_COLUMN].values,

    "Widowed_Population":

        X_test[WIDOWED_COLUMN].values,

    "Actual_Divorced_Population":

        y_test.values,

    "Predicted_Divorced_Population":

        best_predictions

})


result_df["Absolute_Error"] = np.abs(

    result_df["Actual_Divorced_Population"]

    -

    result_df["Predicted_Divorced_Population"]

)


result_df["Percentage_Error"] = np.where(

    result_df["Actual_Divorced_Population"] > 0,

    result_df["Absolute_Error"]

    /

    result_df["Actual_Divorced_Population"].replace(
        0,
        np.nan
    )

    * 100,

    np.nan
)


result_df = result_df.sort_values(
    by="State/UT"
)


result_path = os.path.join(
    BASE_DIR,
    "result.csv"
)


result_df.to_csv(

    result_path,

    index=False
)


print("\nResult CSV Saved:")

print(result_path)


print("\nResult Data:")

print(result_df.head(10))


# ================================================================
# 16. RESULT GRAPH
# ================================================================

print("\nGenerating Result Graph...")


plt.figure(
    figsize=(14, 7)
)


x = np.arange(
    len(result_df)
)


plt.plot(

    x,

    result_df["Actual_Divorced_Population"],

    marker="o",

    linewidth=2.5,

    color="blue",

    label="Actual Population"
)


plt.plot(

    x,

    result_df["Predicted_Divorced_Population"],

    marker="s",

    linewidth=2.5,

    linestyle="--",

    color="red",

    label="Predicted Population"
)


plt.title(

    f"Actual vs Predicted Divorced Women Population ({best_model_name})",

    fontsize=16,

    fontweight="bold"
)


plt.xlabel(
    "States / Union Territories",
    fontsize=12
)


plt.ylabel(
    "Divorced Women Population",
    fontsize=12
)


plt.xticks(

    x,

    result_df["State/UT"],

    rotation=70,

    ha="right"
)


plt.legend()


plt.grid(
    alpha=0.3
)


save_and_show(
    "result_graph.png"
)


# ================================================================
# 17. SAVE BEST MODEL
# ================================================================

print("\nSaving Best Machine Learning Model...")


# Refit best model using complete dataset

best_model.fit(
    X,
    y
)


model_path = os.path.join(
    BASE_DIR,
    "model.pkl"
)


joblib.dump(
    best_model,
    model_path
)


print("Model Saved:", model_path)


# ================================================================
# 18. GENERATE PREDICTION CSV
# ================================================================

print("\nGenerating Prediction CSV...")


# Predictions on all available state records
# using the refitted model

all_predictions = best_model.predict(
    X
)


all_predictions = np.maximum(
    all_predictions,
    0
)


prediction_df = pd.DataFrame({

    "State/UT":

        df[STATE_COLUMN].values,

    "Widowed_Population":

        df[WIDOWED_COLUMN].values,

    "Actual_Divorced_Population":

        df[DIVORCED_COLUMN].values,

    "Predicted_Divorced_Population":

        all_predictions

})


prediction_df["Prediction_Difference"] = (

    prediction_df["Actual_Divorced_Population"]

    -

    prediction_df["Predicted_Divorced_Population"]

)


prediction_df = prediction_df.sort_values(
    by="State/UT"
)


prediction_path = os.path.join(
    BASE_DIR,
    "prediction.csv"
)


prediction_df.to_csv(

    prediction_path,

    index=False
)


print("\nPrediction CSV Saved:")

print(prediction_path)


print("\nPrediction Data:")

print(prediction_df.head(10))


# ================================================================
# 19. PREDICTION GRAPH
# ================================================================

print("\nGenerating Prediction Graph...")


plt.figure(
    figsize=(16, 9)
)


plot_df = prediction_df.sort_values(

    by="Predicted_Divorced_Population",

    ascending=True
)


y_positions = np.arange(
    len(plot_df)
)


plt.barh(

    y_positions,

    plot_df["Predicted_Divorced_Population"],

    color=sns.color_palette(
        "viridis",
        len(plot_df)
    ),

    edgecolor="black"
)


plt.yticks(

    y_positions,

    plot_df["State/UT"],

    fontsize=9
)


plt.xlabel(

    "Predicted Divorced Women Population",

    fontsize=12
)


plt.ylabel(

    "States / Union Territories",

    fontsize=12
)


plt.title(

    "State-Wise Predicted Divorced Women Population",

    fontsize=16,

    fontweight="bold"
)


plt.grid(
    axis="x",
    alpha=0.3
)


save_and_show(
    "prediction_graph.png"
)


# ================================================================
# 20. FINAL MODEL PERFORMANCE
# ================================================================

print("\n" + "=" * 70)

print("FINAL MODEL PERFORMANCE")

print("=" * 70)


best_row = results_df[
    results_df["Model"] == best_model_name
].iloc[0]


print("\nBest Model:", best_model_name)


print(
    "MAE:",
    round(
        float(best_row["MAE"]),
        4
    )
)


print(
    "MSE:",
    round(
        float(best_row["MSE"]),
        4
    )
)


print(
    "RMSE:",
    round(
        float(best_row["RMSE"]),
        4
    )
)


print(
    "R2:",
    round(
        float(best_row["R2"]),
        4
    )
)


print(
    "Cross Validation MAE:",
    round(
        float(best_row["CV_MAE"]),
        4
    )
)


# ================================================================
# 21. VERIFY GENERATED FILES
# ================================================================

print("\n" + "=" * 70)

print("VERIFYING GENERATED FILES")

print("=" * 70)


generated_files = [

    "accuracy_graph.png",

    "heatmap.png",

    "comparison_graph.png",

    "result.csv",

    "result_graph.png",

    "prediction.csv",

    "prediction_graph.png",

    "model.pkl"

]


for filename in generated_files:

    file_path = os.path.join(
        BASE_DIR,
        filename
    )

    if os.path.exists(file_path):

        file_size = os.path.getsize(
            file_path
        ) / 1024

        print(

            f"[SUCCESS] {filename} "

            f"({file_size:.2f} KB)"
        )

    else:

        print(
            f"[FAILED] {filename}"
        )


# ================================================================
# 22. COMPLETION MESSAGE
# ================================================================

print("\n" + "=" * 70)

print("PROJECT EXECUTION COMPLETED SUCCESSFULLY")

print("=" * 70)


print("\nGenerated Files:")

for filename in generated_files:

    print(" -", filename)


print("\nAll files saved in:")

print(BASE_DIR)


print("\nIMPORTANT NOTES:")

print(
    "1. R2 is a regression performance metric, "
    "not classification accuracy."
)

print(
    "2. result.csv contains held-out test predictions."
)

print(
    "3. prediction.csv contains fitted predictions "
    "for all available state records."
)

print(
    "4. These are exploratory estimates, "
    "not future population forecasts."
)

print(
    "5. A small state-level dataset can produce "
    "unstable performance estimates."
)

print("\nThank you!")
```
