```python
import os
import json
import copy
import warnings
import random

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import yaml
import h5py

from sklearn.model_selection import (
    train_test_split,
    KFold,
    cross_val_score
)

from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

# ================================================================
# 1. GLOBAL CONFIGURATION
# ================================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)

sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (12, 7)
plt.rcParams["font.size"] = 11

BASE_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based State-Wise Widowed and Divorced Women Population Analysis"
)

DATASET_PATH = os.path.join(
    BASE_DIR,
    "RS_Session_259_AU_3352_A_and_B.csv"
)

os.makedirs(BASE_DIR, exist_ok=True)

STATE_COLUMN = "State/ UT"

WIDOWED_COLUMN = "Female of all Ages - Widowed"

DIVORCED_COLUMN = "Female of all Ages - Divorced"

# ================================================================
# 2. AIS CONFIGURATION
# ================================================================

POPULATION_SIZE = 12

GENERATIONS = 20

CLONE_FACTOR = 4

ELITE_COUNT = 4

RANDOM_REPLACEMENTS = 3

MUTATION_RATE = 0.30

CV_FOLDS = 5

# Hyperparameter search space

PARAMETER_BOUNDS = {

    "n_estimators": (30, 200),

    "max_depth": (2, 15),

    "min_samples_split": (2, 12),

    "min_samples_leaf": (1, 6)

}

PARAMETER_NAMES = list(
    PARAMETER_BOUNDS.keys()
)

rng = np.random.default_rng(SEED)

# ================================================================
# 3. HELPER FUNCTIONS
# ================================================================

def output_path(filename):

    return os.path.join(
        BASE_DIR,
        "ais_" + filename
    )


def save_and_show(filename):

    path = output_path(filename)

    plt.tight_layout()

    plt.savefig(
        path,
        dpi=300,
        bbox_inches="tight"
    )

    print("[SAVED]", path)

    plt.show()

    plt.close()


def evaluate_predictions(y_true, y_pred):

    y_pred = np.maximum(
        np.asarray(y_pred),
        0
    )

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    mse = mean_squared_error(
        y_true,
        y_pred
    )

    rmse = np.sqrt(mse)

    r2 = r2_score(
        y_true,
        y_pred
    )

    return {

        "MAE": float(mae),

        "MSE": float(mse),

        "RMSE": float(rmse),

        "R2": float(r2)

    }


def create_pipeline(params):

    return Pipeline([

        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),

        (
            "scaler",
            StandardScaler()
        ),

        (
            "regressor",
            RandomForestRegressor(

                n_estimators=int(
                    params["n_estimators"]
                ),

                max_depth=int(
                    params["max_depth"]
                ),

                min_samples_split=int(
                    params["min_samples_split"]
                ),

                min_samples_leaf=int(
                    params["min_samples_leaf"]
                ),

                random_state=SEED,

                n_jobs=-1
            )
        )

    ])


# ================================================================
# 4. LOAD DATASET
# ================================================================

print("=" * 70)

print("AIS-BASED WOMEN POPULATION ANALYSIS")

print("=" * 70)

print("\nLoading Dataset...")

try:

    df = pd.read_csv(
        DATASET_PATH,
        encoding="utf-8-sig"
    )

except UnicodeDecodeError:

    df = pd.read_csv(
        DATASET_PATH,
        encoding="latin1"
    )

df.columns = df.columns.str.strip()

required_columns = [

    STATE_COLUMN,

    WIDOWED_COLUMN,

    DIVORCED_COLUMN

]

missing = [

    col for col in required_columns

    if col not in df.columns

]

if missing:

    raise ValueError(
        f"Missing columns: {missing}\n"
        f"Available columns: {df.columns.tolist()}"
    )

print("\nOriginal Dataset Shape:")

print(df.shape)

print("\nDataset Preview:")

print(df.head())

# ================================================================
# 5. DATA CLEANING
# ================================================================

print("\nCleaning Dataset...")

df = df.drop_duplicates()

df[STATE_COLUMN] = (
    df[STATE_COLUMN]
    .astype("string")
    .str.strip()
)

for col in [
    WIDOWED_COLUMN,
    DIVORCED_COLUMN
]:

    df[col] = pd.to_numeric(

        df[col]
        .astype(str)
        .str.replace(
            ",",
            "",
            regex=False
        ),

        errors="coerce"
    )

df = df.dropna(
    subset=[
        STATE_COLUMN,
        DIVORCED_COLUMN
    ]
)

df = df[
    df[DIVORCED_COLUMN] >= 0
]

df = df.reset_index(drop=True)

if len(df) < 12:

    raise ValueError(
        "Insufficient records for AIS model training."
    )

print("\nCleaned Dataset Shape:")

print(df.shape)

print("\nMissing Values:")

print(df.isnull().sum())

# ================================================================
# 6. FEATURES AND TARGET
# ================================================================

# State names are retained for reporting.
#
# Only the widowed population is used as a numerical predictor.
#
# With approximately one record per state, one-hot encoding
# state names would not generalize well to unseen states.

X = df[[WIDOWED_COLUMN]]

y = df[DIVORCED_COLUMN].astype(float)

X_train, X_test, y_train, y_test = train_test_split(

    X,

    y,

    test_size=0.25,

    random_state=SEED
)

print("\nTraining Records:", len(X_train))

print("Testing Records:", len(X_test))

# ================================================================
# 7. BASELINE RANDOM FOREST
# ================================================================

print("\nTraining Baseline Random Forest...")

baseline_params = {

    "n_estimators": 100,

    "max_depth": 5,

    "min_samples_split": 2,

    "min_samples_leaf": 1

}

baseline_model = create_pipeline(
    baseline_params
)

baseline_model.fit(
    X_train,
    y_train
)

baseline_predictions = np.maximum(

    baseline_model.predict(X_test),

    0
)

baseline_metrics = evaluate_predictions(

    y_test,

    baseline_predictions
)

print("\nBaseline Results:")

for key, value in baseline_metrics.items():

    print(
        key,
        ":",
        round(value, 4)
    )

# ================================================================
# 8. AIS ANTIBODY GENERATION
# ================================================================

def generate_antibody():

    antibody = {}

    for name, bounds in PARAMETER_BOUNDS.items():

        low, high = bounds

        antibody[name] = int(
            rng.integers(
                low,
                high + 1
            )
        )

    return antibody


def antibody_key(antibody):

    return tuple(

        antibody[name]

        for name in PARAMETER_NAMES

    )


# ================================================================
# 9. FITNESS FUNCTION
# ================================================================

fitness_cache = {}

cv = KFold(

    n_splits=CV_FOLDS,

    shuffle=True,

    random_state=SEED
)


def fitness_function(antibody):

    key = antibody_key(
        antibody
    )

    if key in fitness_cache:

        return fitness_cache[key]

    model = create_pipeline(
        antibody
    )

    scores = cross_val_score(

        model,

        X_train,

        y_train,

        cv=cv,

        scoring="neg_mean_absolute_error",

        n_jobs=1
    )

    # Lower MAE means better fitness.
    #
    # The test set is never used during optimization.

    fitness = float(
        -np.mean(scores)
    )

    fitness_cache[key] = fitness

    return fitness


# ================================================================
# 10. AIS CLONING
# ================================================================

def clone_antibody(antibody):

    return copy.deepcopy(
        antibody
    )


# ================================================================
# 11. AIS HYPERMUTATION
# ================================================================

def mutate_antibody(
    antibody,
    mutation_rate
):

    mutated = clone_antibody(
        antibody
    )

    for name, bounds in PARAMETER_BOUNDS.items():

        if rng.random() < mutation_rate:

            low, high = bounds

            span = high - low

            change = int(
                rng.normal(
                    0,
                    max(1, span * 0.15)
                )
            )

            mutated[name] = int(
                np.clip(

                    mutated[name] + change,

                    low,

                    high
                )
            )

    return mutated


# ================================================================
# 12. INITIALIZE AIS POPULATION
# ================================================================

print("\nInitializing AIS Population...")

population = [

    generate_antibody()

    for _ in range(POPULATION_SIZE)

]

best_antibody = None

best_fitness = float("inf")

fitness_history = []

generation_best_history = []

mean_fitness_history = []

# ================================================================
# 13. AIS CLONAL SELECTION ALGORITHM
# ================================================================

print("\nStarting AIS Optimization...")

for generation in range(GENERATIONS):

    print("\n" + "=" * 60)

    print(
        f"AIS Generation {generation + 1}/{GENERATIONS}"
    )

    print("=" * 60)

    # Evaluate population

    evaluated_population = [

        (
            fitness_function(antibody),
            antibody
        )

        for antibody in population

    ]

    evaluated_population.sort(
        key=lambda item: item[0]
    )

    generation_best_fitness = (
        evaluated_population[0][0]
    )

    generation_best_antibody = (
        evaluated_population[0][1]
    )

    generation_mean_fitness = float(
        np.mean(
            [
                item[0]
                for item in evaluated_population
            ]
        )
    )

    # Update global best

    if generation_best_fitness < best_fitness:

        best_fitness = generation_best_fitness

        best_antibody = clone_antibody(
            generation_best_antibody
        )

    fitness_history.append(
        best_fitness
    )

    generation_best_history.append(
        generation_best_fitness
    )

    mean_fitness_history.append(
        generation_mean_fitness
    )

    print(
        "Generation Best CV MAE:",
        round(generation_best_fitness, 4)
    )

    print(
        "Global Best CV MAE:",
        round(best_fitness, 4)
    )

    print(
        "Best Parameters:",
        best_antibody
    )

    # Select elite antibodies

    elite_antibodies = [

        clone_antibody(item[1])

        for item in evaluated_population[
            :ELITE_COUNT
        ]

    ]

    # Clone and mutate elite antibodies

    clones = []

    for rank, antibody in enumerate(
        elite_antibodies
    ):

        number_of_clones = max(

            1,

            int(
                CLONE_FACTOR
                * ELITE_COUNT
                / (rank + 1)
            )
        )

        # Higher-ranked antibodies undergo
        # lower mutation intensity.

        adaptive_mutation_rate = min(

            1.0,

            MUTATION_RATE
            * (rank + 1)
        )

        for _ in range(number_of_clones):

            clone = clone_antibody(
                antibody
            )

            clone = mutate_antibody(

                clone,

                adaptive_mutation_rate
            )

            clones.append(
                clone
            )

    # Evaluate clones

    evaluated_clones = [

        (
            fitness_function(clone),
            clone
        )

        for clone in clones

    ]

    # Combine elite population and clones

    combined = (
        evaluated_population
        +
        evaluated_clones
    )

    combined.sort(
        key=lambda item: item[0]
    )

    # Keep best unique antibodies

    next_population = []

    seen = set()

    for fitness, antibody in combined:

        key = antibody_key(
            antibody
        )

        if key not in seen:

            next_population.append(
                clone_antibody(antibody)
            )

            seen.add(key)

        if len(next_population) >= (
            POPULATION_SIZE
            -
            RANDOM_REPLACEMENTS
        ):

            break

    # Introduce random antibodies
    # to maintain population diversity

    while len(next_population) < POPULATION_SIZE:

        new_antibody = generate_antibody()

        key = antibody_key(
            new_antibody
        )

        if key not in seen:

            next_population.append(
                new_antibody
            )

            seen.add(key)

    population = next_population


print("\n" + "=" * 70)

print("AIS OPTIMIZATION COMPLETED")

print("=" * 70)

print("\nBest Parameters:")

print(best_antibody)

print("\nBest Cross-Validation MAE:")

print(best_fitness)

# ================================================================
# 14. TRAIN AIS-OPTIMIZED MODEL
# ================================================================

print("\nTraining AIS-Optimized Random Forest...")

ais_model = create_pipeline(
    best_antibody
)

ais_model.fit(
    X_train,
    y_train
)

ais_predictions = np.maximum(

    ais_model.predict(
        X_test
    ),

    0
)

ais_metrics = evaluate_predictions(

    y_test,

    ais_predictions
)

print("\nAIS Model Evaluation:")

for key, value in ais_metrics.items():

    print(
        key,
        ":",
        round(value, 4)
    )

# ================================================================
# 15. AIS ACCURACY GRAPH
# ================================================================

# This is an optimization convergence graph.
# Regression does not have classification accuracy.

print("\nGenerating AIS Accuracy Graph...")

generations = np.arange(
    1,
    GENERATIONS + 1
)

plt.figure(
    figsize=(12, 7)
)

plt.plot(

    generations,

    fitness_history,

    marker="o",

    linewidth=2.5,

    color="blue",

    label="Global Best CV MAE"
)

plt.plot(

    generations,

    generation_best_history,

    marker="s",

    linestyle="--",

    linewidth=1.8,

    color="green",

    label="Generation Best CV MAE"
)

plt.plot(

    generations,

    mean_fitness_history,

    linestyle=":",

    linewidth=2,

    color="orange",

    label="Mean Population CV MAE"
)

plt.title(

    "AIS Optimization Convergence",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel(
    "AIS Generation"
)

plt.ylabel(
    "Cross-Validation MAE"
)

plt.legend()

plt.grid(
    alpha=0.3
)

save_and_show(
    "accuracy_graph.png"
)

# ================================================================
# 16. AIS HEATMAP
# ================================================================

print("\nGenerating AIS Heatmap...")

correlation_data = df[
    [
        WIDOWED_COLUMN,
        DIVORCED_COLUMN
    ]
]

correlation_matrix = (
    correlation_data.corr()
)

plt.figure(
    figsize=(10, 7)
)

sns.heatmap(

    correlation_matrix,

    annot=True,

    fmt=".3f",

    cmap="coolwarm",

    linewidths=1,

    square=True,

    vmin=-1,

    vmax=1,

    cbar_kws={
        "label": "Correlation Coefficient"
    }
)

plt.title(

    "AIS Project - Population Correlation Heatmap",

    fontsize=15,

    fontweight="bold"
)

plt.xticks(
    rotation=20,
    ha="right"
)

plt.yticks(
    rotation=0
)

save_and_show(
    "heatmap.png"
)

# ================================================================
# 17. AIS COMPARISON GRAPH
# ================================================================

print("\nGenerating AIS Comparison Graph...")

comparison_data = pd.DataFrame({

    "Model": [

        "Baseline Random Forest",

        "AIS Optimized Random Forest"

    ],

    "MAE": [

        baseline_metrics["MAE"],

        ais_metrics["MAE"]

    ],

    "RMSE": [

        baseline_metrics["RMSE"],

        ais_metrics["RMSE"]

    ],

    "R2": [

        baseline_metrics["R2"],

        ais_metrics["R2"]

    ]

})

x = np.arange(
    len(comparison_data)
)

width = 0.35

plt.figure(
    figsize=(12, 7)
)

plt.bar(

    x - width / 2,

    comparison_data["MAE"],

    width,

    label="MAE",

    color="steelblue",

    edgecolor="black"
)

plt.bar(

    x + width / 2,

    comparison_data["RMSE"],

    width,

    label="RMSE",

    color="coral",

    edgecolor="black"
)

plt.xticks(

    x,

    comparison_data["Model"],

    rotation=10
)

plt.xlabel(
    "Machine Learning Model"
)

plt.ylabel(
    "Prediction Error (Women)"
)

plt.title(

    "Baseline vs AIS-Optimized Random Forest",

    fontsize=16,

    fontweight="bold"
)

plt.legend()

plt.grid(
    axis="y",
    alpha=0.3
)

save_and_show(
    "comparison_graph.png"
)

# ================================================================
# 18. GENERATE AIS RESULT CSV
# ================================================================

print("\nGenerating AIS Result CSV...")

result_df = pd.DataFrame({

    "State/UT":

        df.loc[
            X_test.index,
            STATE_COLUMN
        ].values,

    "Widowed_Population":

        X_test[
            WIDOWED_COLUMN
        ].values,

    "Actual_Divorced_Population":

        y_test.values,

    "AIS_Predicted_Divorced_Population":

        ais_predictions

})

result_df["Absolute_Error"] = np.abs(

    result_df["Actual_Divorced_Population"]

    -

    result_df["AIS_Predicted_Divorced_Population"]

)

result_df["Percentage_Error"] = (

    result_df["Absolute_Error"]

    /

    result_df["Actual_Divorced_Population"].replace(
        0,
        np.nan
    )

    * 100
)

result_df = result_df.sort_values(
    by="State/UT"
)

result_df.to_csv(

    output_path(
        "result.csv"
    ),

    index=False
)

print("\nAIS Result Data:")

print(result_df.head(10))

# ================================================================
# 19. AIS RESULT GRAPH
# ================================================================

print("\nGenerating AIS Result Graph...")

plt.figure(
    figsize=(14, 7)
)

x = np.arange(
    len(result_df)
)

plt.plot(

    x,

    result_df[
        "Actual_Divorced_Population"
    ],

    marker="o",

    linewidth=2.5,

    label="Actual Population",

    color="blue"
)

plt.plot(

    x,

    result_df[
        "AIS_Predicted_Divorced_Population"
    ],

    marker="s",

    linewidth=2.5,

    linestyle="--",

    label="AIS Predicted Population",

    color="red"
)

plt.xticks(

    x,

    result_df["State/UT"],

    rotation=70,

    ha="right"
)

plt.xlabel(
    "States / Union Territories"
)

plt.ylabel(
    "Divorced Women Population"
)

plt.title(

    "AIS Actual vs Predicted Divorced Women Population",

    fontsize=16,

    fontweight="bold"
)

plt.legend()

plt.grid(
    alpha=0.3
)

save_and_show(
    "result_graph.png"
)

# ================================================================
# 20. RETRAIN AIS MODEL ON COMPLETE DATASET
# ================================================================

print("\nRetraining AIS Model on Complete Dataset...")

final_ais_model = create_pipeline(
    best_antibody
)

final_ais_model.fit(
    X,
    y
)

joblib.dump(

    final_ais_model,

    output_path(
        "model.pkl"
    )
)

print(
    "AIS Model Saved:",
    output_path("model.pkl")
)

# ================================================================
# 21. GENERATE AIS PREDICTION CSV
# ================================================================

print("\nGenerating AIS Prediction CSV...")

all_predictions = np.maximum(

    final_ais_model.predict(X),

    0
)

prediction_df = pd.DataFrame({

    "State/UT":

        df[STATE_COLUMN].values,

    "Widowed_Population":

        df[WIDOWED_COLUMN].values,

    "Actual_Divorced_Population":

        df[DIVORCED_COLUMN].values,

    "AIS_Predicted_Divorced_Population":

        all_predictions

})

prediction_df["Prediction_Difference"] = (

    prediction_df["Actual_Divorced_Population"]

    -

    prediction_df["AIS_Predicted_Divorced_Population"]

)

prediction_df = prediction_df.sort_values(
    by="State/UT"
)

prediction_df.to_csv(

    output_path(
        "prediction.csv"
    ),

    index=False
)

print("\nAIS Prediction Data:")

print(prediction_df.head(10))

# ================================================================
# 22. AIS PREDICTION GRAPH
# ================================================================

print("\nGenerating AIS Prediction Graph...")

plot_df = prediction_df.sort_values(

    by="AIS_Predicted_Divorced_Population",

    ascending=True
)

plt.figure(
    figsize=(16, 10)
)

y_positions = np.arange(
    len(plot_df)
)

plt.barh(

    y_positions,

    plot_df[
        "AIS_Predicted_Divorced_Population"
    ],

    color=sns.color_palette(
        "viridis",
        len(plot_df)
    ),

    edgecolor="black"
)

plt.yticks(

    y_positions,

    plot_df["State/UT"],

    fontsize=9
)

plt.xlabel(
    "AIS Predicted Divorced Women Population"
)

plt.ylabel(
    "States / Union Territories"
)

plt.title(

    "AIS State-Wise Predicted Divorced Women Population",

    fontsize=16,

    fontweight="bold"
)

plt.grid(
    axis="x",
    alpha=0.3
)

save_and_show(
    "prediction_graph.png"
)

# ================================================================
# 23. SAVE AIS MODEL JSON
# ================================================================

print("\nSaving AIS Model JSON...")

json_data = {

    "project_name":
        "AI-Based State-Wise Widowed and Divorced Women Population Analysis",

    "optimization_algorithm":
        "Artificial Immune System",

    "ais_variant":
        "Clonal Selection Algorithm",

    "machine_learning_model":
        "Random Forest Regressor",

    "dataset": {

        "filename":
            os.path.basename(DATASET_PATH),

        "total_records":
            int(len(df)),

        "input_feature":
            WIDOWED_COLUMN,

        "target":
            DIVORCED_COLUMN

    },

    "ais_configuration": {

        "population_size":
            POPULATION_SIZE,

        "generations":
            GENERATIONS,

        "clone_factor":
            CLONE_FACTOR,

        "elite_count":
            ELITE_COUNT,

        "random_replacements":
            RANDOM_REPLACEMENTS,

        "mutation_rate":
            MUTATION_RATE,

        "cv_folds":
            CV_FOLDS

    },

    "best_hyperparameters":
        best_antibody,

    "best_cross_validation_mae":
        float(best_fitness),

    "baseline_metrics":
        baseline_metrics,

    "ais_metrics":
        ais_metrics,

    "optimization_history": {

        "global_best_cv_mae":
            fitness_history,

        "generation_best_cv_mae":
            generation_best_history,

        "mean_population_cv_mae":
            mean_fitness_history

    },

    "model_file":
        "ais_model.pkl"

}

with open(

    output_path(
        "model.json"
    ),

    "w",

    encoding="utf-8"

) as file:

    json.dump(

        json_data,

        file,

        indent=4
    )

print(
    "Saved:",
    output_path("model.json")
)

# ================================================================
# 24. SAVE AIS MODEL YAML
# ================================================================

print("\nSaving AIS Model YAML...")

yaml_data = {

    "project": {

        "name":
            "AI-Based State-Wise Widowed and Divorced Women Population Analysis",

        "algorithm":
            "Artificial Immune System",

        "variant":
            "Clonal Selection Algorithm",

        "model":
            "Random Forest Regressor"

    },

    "dataset": {

        "records":
            int(len(df)),

        "input_feature":
            WIDOWED_COLUMN,

        "target":
            DIVORCED_COLUMN

    },

    "ais_configuration": {

        "population_size":
            POPULATION_SIZE,

        "generations":
            GENERATIONS,

        "clone_factor":
            CLONE_FACTOR,

        "mutation_rate":
            MUTATION_RATE

    },

    "best_parameters":
        best_antibody,

    "best_cv_mae":
        float(best_fitness),

    "evaluation": {

        "baseline":
            baseline_metrics,

        "ais_optimized":
            ais_metrics

    },

    "saved_model":
        "ais_model.pkl"

}

with open(

    output_path(
        "model.yaml"
    ),

    "w",

    encoding="utf-8"

) as file:

    yaml.safe_dump(

        yaml_data,

        file,

        sort_keys=False
    )

print(
    "Saved:",
    output_path("model.yaml")
)

# ================================================================
# 25. SAVE AIS H5 FILE
# ================================================================

print("\nSaving AIS H5 File...")

# H5 stores AIS optimization results and metadata.
# The fitted Random Forest is stored separately in PKL.

with h5py.File(

    output_path(
        "model.h5"
    ),

    "w"

) as h5_file:

    h5_file.attrs[
        "algorithm"
    ] = "Artificial Immune System"

    h5_file.attrs[
        "variant"
    ] = "Clonal Selection Algorithm"

    h5_file.attrs[
        "model"
    ] = "Random Forest Regressor"

    h5_file.attrs[
        "best_cv_mae"
    ] = float(best_fitness)

    h5_file.create_dataset(

        "global_best_cv_mae",

        data=np.array(
            fitness_history
        )
    )

    h5_file.create_dataset(

        "generation_best_cv_mae",

        data=np.array(
            generation_best_history
        )
    )

    h5_file.create_dataset(

        "mean_population_cv_mae",

        data=np.array(
            mean_fitness_history
        )
    )

    parameters_group = (
        h5_file.create_group(
            "best_parameters"
        )
    )

    for key, value in best_antibody.items():

        parameters_group.attrs[key] = int(
            value
        )

    metrics_group = (
        h5_file.create_group(
            "evaluation_metrics"
        )
    )

    for key, value in ais_metrics.items():

        metrics_group.attrs[key] = float(
            value
        )

print(
    "Saved:",
    output_path("model.h5")
)

# ================================================================
# 26. VERIFY ALL OUTPUT FILES
# ================================================================

print("\n" + "=" * 70)

print("VERIFYING AIS OUTPUT FILES")

print("=" * 70)

generated_files = [

    "ais_model.pkl",

    "ais_model.h5",

    "ais_model.yaml",

    "ais_model.json",

    "ais_accuracy_graph.png",

    "ais_heatmap.png",

    "ais_comparison_graph.png",

    "ais_result.csv",

    "ais_result_graph.png",

    "ais_prediction.csv",

    "ais_prediction_graph.png"

]

for filename in generated_files:

    path = os.path.join(
        BASE_DIR,
        filename
    )

    if os.path.exists(path):

        size = os.path.getsize(
            path
        ) / 1024

        print(
            f"[SUCCESS] {filename} ({size:.2f} KB)"
        )

    else:

        print(
            f"[FAILED] {filename}"
        )

# ================================================================
# 27. FINAL SUMMARY
# ================================================================

print("\n" + "=" * 70)

print("AIS PROJECT EXECUTION COMPLETED")

print("=" * 70)

print("\nOptimization Algorithm:")
print("Artificial Immune System - CLONALG")

print("\nMachine Learning Algorithm:")
print("Random Forest Regressor")

print("\nBest Hyperparameters:")
print(best_antibody)

print("\nBest Cross-Validation MAE:")
print(round(best_fitness, 4))

print("\nBaseline Performance:")

for metric, value in baseline_metrics.items():

    print(
        metric,
        ":",
        round(value, 4)
    )

print("\nAIS Optimized Performance:")

for metric, value in ais_metrics.items():

    print(
        metric,
        ":",
        round(value, 4)
    )

print("\nAll Files Saved In:")

print(BASE_DIR)

print("\nIMPORTANT:")

print(
    "AIS optimization minimizes cross-validation MAE."
)

print(
    "Lower MAE and RMSE indicate better regression performance."
)

print(
    "R2 measures regression fit, not classification accuracy."
)

print(
    "Test predictions are held out from training and optimization."
)

print(
    "Full-dataset predictions are fitted estimates, "
    "not future forecasts."
)

print(
    "AIS does not guarantee better test performance "
    "than the baseline model."
)

print("\nThank You!")
```
