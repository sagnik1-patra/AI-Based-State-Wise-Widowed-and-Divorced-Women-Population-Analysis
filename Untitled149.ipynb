```python
import os
import json
import random
import warnings

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

import joblib
import yaml
import h5py

from sklearn.model_selection import (
    train_test_split,
    KFold,
    cross_val_score
)

from sklearn.pipeline import Pipeline

from sklearn.preprocessing import StandardScaler

from sklearn.impute import SimpleImputer

from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")


# ================================================================
# 2. GLOBAL SETTINGS
# ================================================================

SEED = 42

random.seed(SEED)

np.random.seed(SEED)

rng = np.random.default_rng(SEED)

sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (12, 7)

plt.rcParams["font.size"] = 11


# ================================================================
# 3. PROJECT DIRECTORY
# ================================================================

BASE_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based State-Wise Widowed and Divorced Women Population Analysis"
)

DATASET_PATH = os.path.join(
    BASE_DIR,
    "RS_Session_259_AU_3352_A_and_B.csv"
)

os.makedirs(BASE_DIR, exist_ok=True)


STATE_COLUMN = "State/ UT"

WIDOWED_COLUMN = "Female of all Ages - Widowed"

DIVORCED_COLUMN = "Female of all Ages - Divorced"


# ================================================================
# 4. PSO CONFIGURATION
# ================================================================

NUM_PARTICLES = 12

MAX_ITERATIONS = 20

INERTIA_WEIGHT = 0.7

COGNITIVE_COEFFICIENT = 1.5

SOCIAL_COEFFICIENT = 1.5

CV_FOLDS = 5


# Random Forest hyperparameter search space

PARAMETER_BOUNDS = {

    "n_estimators": (30, 200),

    "max_depth": (2, 15),

    "min_samples_split": (2, 12),

    "min_samples_leaf": (1, 6)

}

PARAMETER_NAMES = list(
    PARAMETER_BOUNDS.keys()
)

LOWER_BOUNDS = np.array(
    [
        PARAMETER_BOUNDS[name][0]
        for name in PARAMETER_NAMES
    ],
    dtype=float
)

UPPER_BOUNDS = np.array(
    [
        PARAMETER_BOUNDS[name][1]
        for name in PARAMETER_NAMES
    ],
    dtype=float
)

DIMENSIONS = len(PARAMETER_NAMES)


# ================================================================
# 5. HELPER FUNCTIONS
# ================================================================

def output_path(filename):

    return os.path.join(
        BASE_DIR,
        "pso_" + filename
    )


def save_and_show(filename):

    path = output_path(filename)

    plt.tight_layout()

    plt.savefig(
        path,
        dpi=300,
        bbox_inches="tight"
    )

    print("\n[SAVED]", path)

    plt.show()

    plt.close()


def evaluate_predictions(y_true, y_pred):

    y_pred = np.maximum(
        np.asarray(y_pred, dtype=float),
        0
    )

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    mse = mean_squared_error(
        y_true,
        y_pred
    )

    rmse = np.sqrt(mse)

    r2 = r2_score(
        y_true,
        y_pred
    )

    return {

        "MAE": float(mae),

        "MSE": float(mse),

        "RMSE": float(rmse),

        "R2": float(r2)

    }


def position_to_parameters(position):

    """
    Convert continuous PSO particle positions
    into valid integer Random Forest parameters.
    """

    position = np.clip(
        position,
        LOWER_BOUNDS,
        UPPER_BOUNDS
    )

    values = np.rint(position).astype(int)

    parameters = {

        name: int(value)

        for name, value in zip(
            PARAMETER_NAMES,
            values
        )

    }

    return parameters


def create_pipeline(params):

    """
    Create Random Forest regression pipeline.
    """

    return Pipeline([

        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),

        (
            "scaler",
            StandardScaler()
        ),

        (
            "regressor",
            RandomForestRegressor(

                n_estimators=int(
                    params["n_estimators"]
                ),

                max_depth=int(
                    params["max_depth"]
                ),

                min_samples_split=int(
                    params["min_samples_split"]
                ),

                min_samples_leaf=int(
                    params["min_samples_leaf"]
                ),

                random_state=SEED,

                n_jobs=-1
            )
        )

    ])


# ================================================================
# 6. LOAD DATASET
# ================================================================

print("=" * 70)

print("PSO-BASED WOMEN POPULATION ANALYSIS")

print("=" * 70)

print("\nLoading Dataset...")

if not os.path.exists(DATASET_PATH):

    raise FileNotFoundError(
        f"Dataset not found: {DATASET_PATH}"
    )

try:

    df = pd.read_csv(
        DATASET_PATH,
        encoding="utf-8-sig"
    )

except UnicodeDecodeError:

    df = pd.read_csv(
        DATASET_PATH,
        encoding="latin1"
    )

df.columns = df.columns.str.strip()

print("\nDataset Loaded Successfully!")

print("\nOriginal Dataset Shape:")

print(df.shape)

print("\nDataset Columns:")

print(df.columns.tolist())

print("\nFirst Five Records:")

print(df.head())


# ================================================================
# 7. VALIDATE DATASET
# ================================================================

required_columns = [

    STATE_COLUMN,

    WIDOWED_COLUMN,

    DIVORCED_COLUMN

]

missing_columns = [

    col for col in required_columns

    if col not in df.columns

]

if missing_columns:

    raise ValueError(
        f"Missing columns: {missing_columns}\n"
        f"Available columns: {df.columns.tolist()}"
    )


# ================================================================
# 8. DATA PREPROCESSING
# ================================================================

print("\nPerforming Data Preprocessing...")

df = df.drop_duplicates()

df[STATE_COLUMN] = (
    df[STATE_COLUMN]
    .astype("string")
    .str.strip()
)

for column in [
    WIDOWED_COLUMN,
    DIVORCED_COLUMN
]:

    df[column] = pd.to_numeric(

        df[column]
        .astype(str)
        .str.replace(
            ",",
            "",
            regex=False
        ),

        errors="coerce"
    )

df = df.dropna(
    subset=[
        STATE_COLUMN,
        DIVORCED_COLUMN
    ]
)

df = df[
    df[DIVORCED_COLUMN] >= 0
]

df = df.reset_index(drop=True)

if len(df) < 12:

    raise ValueError(
        "Insufficient valid records for PSO model training."
    )

print("\nCleaned Dataset Shape:")

print(df.shape)

print("\nMissing Values:")

print(df.isnull().sum())

print("\nStatistical Summary:")

print(df.describe())


# ================================================================
# 9. DEFINE FEATURES AND TARGET
# ================================================================

print("\nPreparing Features and Target...")

# State names are used for reporting.
#
# The widowed population is used as the numerical
# predictor of divorced population.
#
# This avoids using one-hot encoded state identifiers
# that have little generalization value with one
# observation per state.

X = df[
    [
        WIDOWED_COLUMN
    ]
]

y = df[
    DIVORCED_COLUMN
].astype(float)


# ================================================================
# 10. TRAIN TEST SPLIT
# ================================================================

X_train, X_test, y_train, y_test = train_test_split(

    X,

    y,

    test_size=0.25,

    random_state=SEED
)

print("\nTraining Records:", len(X_train))

print("Testing Records:", len(X_test))


# ================================================================
# 11. BASELINE RANDOM FOREST MODEL
# ================================================================

print("\nTraining Baseline Random Forest...")

baseline_params = {

    "n_estimators": 100,

    "max_depth": 5,

    "min_samples_split": 2,

    "min_samples_leaf": 1

}

baseline_model = create_pipeline(
    baseline_params
)

baseline_model.fit(
    X_train,
    y_train
)

baseline_predictions = np.maximum(

    baseline_model.predict(X_test),

    0
)

baseline_metrics = evaluate_predictions(

    y_test,

    baseline_predictions
)

print("\nBaseline Model Performance:")

for metric, value in baseline_metrics.items():

    print(
        f"{metric}: {value:.4f}"
    )


# ================================================================
# 12. PSO FITNESS FUNCTION
# ================================================================

print("\nPreparing PSO Fitness Function...")

cv = KFold(

    n_splits=CV_FOLDS,

    shuffle=True,

    random_state=SEED
)

fitness_cache = {}


def fitness_function(position):

    """
    Fitness = Mean Cross-Validation MAE.

    Lower MAE is better.

    The held-out test set is not used
    during PSO optimization.
    """

    params = position_to_parameters(
        position
    )

    key = tuple(
        params[name]
        for name in PARAMETER_NAMES
    )

    if key in fitness_cache:

        return fitness_cache[key]

    model = create_pipeline(
        params
    )

    scores = cross_val_score(

        model,

        X_train,

        y_train,

        cv=cv,

        scoring="neg_mean_absolute_error",

        n_jobs=1
    )

    fitness = float(
        -np.mean(scores)
    )

    fitness_cache[key] = fitness

    return fitness


# ================================================================
# 13. INITIALIZE PSO PARTICLES
# ================================================================

print("\nInitializing PSO Particles...")

positions = rng.uniform(

    LOWER_BOUNDS,

    UPPER_BOUNDS,

    size=(
        NUM_PARTICLES,
        DIMENSIONS
    )
)

velocity_limits = (
    UPPER_BOUNDS - LOWER_BOUNDS
) * 0.2

velocities = rng.uniform(

    -velocity_limits,

    velocity_limits,

    size=(
        NUM_PARTICLES,
        DIMENSIONS
    )
)

# Personal best positions

personal_best_positions = positions.copy()

personal_best_scores = np.full(
    NUM_PARTICLES,
    np.inf
)

# Global best position

global_best_position = None

global_best_score = float("inf")

# Optimization history

global_best_history = []

iteration_best_history = []

mean_fitness_history = []


# ================================================================
# 14. PARTICLE SWARM OPTIMIZATION
# ================================================================

print("\nStarting Particle Swarm Optimization...")

for iteration in range(MAX_ITERATIONS):

    print("\n" + "=" * 65)

    print(
        f"PSO Iteration {iteration + 1}/{MAX_ITERATIONS}"
    )

    print("=" * 65)

    iteration_scores = []

    # Evaluate each particle

    for particle_index in range(NUM_PARTICLES):

        current_position = positions[
            particle_index
        ]

        current_fitness = fitness_function(
            current_position
        )

        iteration_scores.append(
            current_fitness
        )

        # Update personal best

        if current_fitness < personal_best_scores[
            particle_index
        ]:

            personal_best_scores[
                particle_index
            ] = current_fitness

            personal_best_positions[
                particle_index
            ] = current_position.copy()

        # Update global best

        if current_fitness < global_best_score:

            global_best_score = current_fitness

            global_best_position = (
                current_position.copy()
            )

    iteration_best = float(
        np.min(iteration_scores)
    )

    iteration_mean = float(
        np.mean(iteration_scores)
    )

    global_best_history.append(
        global_best_score
    )

    iteration_best_history.append(
        iteration_best
    )

    mean_fitness_history.append(
        iteration_mean
    )

    print(
        "Iteration Best CV MAE:",
        round(iteration_best, 4)
    )

    print(
        "Global Best CV MAE:",
        round(global_best_score, 4)
    )

    print(
        "Best Parameters:",
        position_to_parameters(
            global_best_position
        )
    )

    # Update particle velocities and positions

    for particle_index in range(NUM_PARTICLES):

        r1 = rng.random(
            DIMENSIONS
        )

        r2 = rng.random(
            DIMENSIONS
        )

        cognitive_component = (

            COGNITIVE_COEFFICIENT

            * r1

            * (
                personal_best_positions[
                    particle_index
                ]

                -

                positions[
                    particle_index
                ]
            )
        )

        social_component = (

            SOCIAL_COEFFICIENT

            * r2

            * (
                global_best_position

                -

                positions[
                    particle_index
                ]
            )
        )

        velocities[
            particle_index
        ] = (

            INERTIA_WEIGHT

            * velocities[
                particle_index
            ]

            +

            cognitive_component

            +

            social_component
        )

        # Limit velocity

        velocities[
            particle_index
        ] = np.clip(

            velocities[
                particle_index
            ],

            -velocity_limits,

            velocity_limits
        )

        # Update position

        positions[
            particle_index
        ] += velocities[
            particle_index
        ]

        # Enforce parameter bounds

        positions[
            particle_index
        ] = np.clip(

            positions[
                particle_index
            ],

            LOWER_BOUNDS,

            UPPER_BOUNDS
        )


# ================================================================
# 15. PSO OPTIMIZATION RESULTS
# ================================================================

best_params = position_to_parameters(
    global_best_position
)

print("\n" + "=" * 70)

print("PSO OPTIMIZATION COMPLETED")

print("=" * 70)

print("\nBest Hyperparameters:")

print(best_params)

print("\nBest Cross-Validation MAE:")

print(global_best_score)

print("\nUnique Parameter Combinations Evaluated:")

print(len(fitness_cache))


# ================================================================
# 16. TRAIN PSO-OPTIMIZED RANDOM FOREST
# ================================================================

print("\nTraining PSO-Optimized Random Forest...")

pso_model = create_pipeline(
    best_params
)

pso_model.fit(
    X_train,
    y_train
)

pso_predictions = np.maximum(

    pso_model.predict(
        X_test
    ),

    0
)

pso_metrics = evaluate_predictions(

    y_test,

    pso_predictions
)

print("\nPSO Model Performance:")

for metric, value in pso_metrics.items():

    print(
        f"{metric}: {value:.4f}"
    )


# ================================================================
# 17. PSO ACCURACY GRAPH
# ================================================================

print("\nGenerating PSO Accuracy Graph...")

# Regression models do not have classification accuracy.
# This graph shows PSO convergence using CV MAE.

iterations = np.arange(
    1,
    MAX_ITERATIONS + 1
)

plt.figure(
    figsize=(12, 7)
)

plt.plot(

    iterations,

    global_best_history,

    marker="o",

    linewidth=2.5,

    color="blue",

    label="Global Best CV MAE"
)

plt.plot(

    iterations,

    iteration_best_history,

    marker="s",

    linestyle="--",

    linewidth=2,

    color="green",

    label="Iteration Best CV MAE"
)

plt.plot(

    iterations,

    mean_fitness_history,

    linestyle=":",

    linewidth=2,

    color="orange",

    label="Mean Particle CV MAE"
)

plt.title(

    "PSO Optimization Convergence",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel(
    "PSO Iteration",
    fontsize=12
)

plt.ylabel(
    "Cross-Validation MAE",
    fontsize=12
)

plt.legend()

plt.grid(
    alpha=0.3
)

save_and_show(
    "accuracy_graph.png"
)


# ================================================================
# 18. PSO CORRELATION HEATMAP
# ================================================================

print("\nGenerating PSO Heatmap...")

numeric_df = df[
    [
        WIDOWED_COLUMN,
        DIVORCED_COLUMN
    ]
]

correlation_matrix = numeric_df.corr()

plt.figure(
    figsize=(10, 7)
)

sns.heatmap(

    correlation_matrix,

    annot=True,

    fmt=".3f",

    cmap="coolwarm",

    linewidths=1,

    linecolor="white",

    square=True,

    vmin=-1,

    vmax=1,

    cbar_kws={
        "label": "Correlation Coefficient"
    }
)

plt.title(

    "PSO Project - Population Correlation Heatmap",

    fontsize=15,

    fontweight="bold"
)

plt.xticks(
    rotation=20,
    ha="right"
)

plt.yticks(
    rotation=0
)

save_and_show(
    "heatmap.png"
)


# ================================================================
# 19. PSO MODEL COMPARISON GRAPH
# ================================================================

print("\nGenerating PSO Comparison Graph...")

comparison_df = pd.DataFrame({

    "Model": [

        "Baseline Random Forest",

        "PSO Optimized Random Forest"

    ],

    "MAE": [

        baseline_metrics["MAE"],

        pso_metrics["MAE"]

    ],

    "RMSE": [

        baseline_metrics["RMSE"],

        pso_metrics["RMSE"]

    ],

    "R2": [

        baseline_metrics["R2"],

        pso_metrics["R2"]

    ]

})

x = np.arange(
    len(comparison_df)
)

width = 0.35

plt.figure(
    figsize=(12, 7)
)

bars1 = plt.bar(

    x - width / 2,

    comparison_df["MAE"],

    width,

    label="MAE",

    color="steelblue",

    edgecolor="black"
)

bars2 = plt.bar(

    x + width / 2,

    comparison_df["RMSE"],

    width,

    label="RMSE",

    color="coral",

    edgecolor="black"
)

plt.xticks(

    x,

    comparison_df["Model"],

    rotation=10
)

plt.xlabel(
    "Machine Learning Models"
)

plt.ylabel(
    "Prediction Error (Women)"
)

plt.title(

    "Baseline vs PSO-Optimized Random Forest",

    fontsize=16,

    fontweight="bold"
)

plt.legend()

plt.grid(
    axis="y",
    alpha=0.3
)

save_and_show(
    "comparison_graph.png"
)


# ================================================================
# 20. GENERATE PSO RESULT CSV
# ================================================================

print("\nGenerating PSO Result CSV...")

result_df = pd.DataFrame({

    "State/UT":

        df.loc[
            X_test.index,
            STATE_COLUMN
        ].values,

    "Widowed_Population":

        X_test[
            WIDOWED_COLUMN
        ].values,

    "Actual_Divorced_Population":

        y_test.values,

    "PSO_Predicted_Divorced_Population":

        pso_predictions

})

result_df["Absolute_Error"] = np.abs(

    result_df["Actual_Divorced_Population"]

    -

    result_df["PSO_Predicted_Divorced_Population"]

)

result_df["Percentage_Error"] = (

    result_df["Absolute_Error"]

    /

    result_df["Actual_Divorced_Population"].replace(
        0,
        np.nan
    )

    * 100
)

result_df = result_df.sort_values(
    by="State/UT"
)

result_df.to_csv(

    output_path(
        "result.csv"
    ),

    index=False
)

print("\nPSO Result CSV Saved:")

print(
    output_path("result.csv")
)

print("\nResult Preview:")

print(result_df.head(10))


# ================================================================
# 21. PSO RESULT GRAPH
# ================================================================

print("\nGenerating PSO Result Graph...")

plt.figure(
    figsize=(14, 7)
)

x = np.arange(
    len(result_df)
)

plt.plot(

    x,

    result_df[
        "Actual_Divorced_Population"
    ],

    marker="o",

    linewidth=2.5,

    color="blue",

    label="Actual Population"
)

plt.plot(

    x,

    result_df[
        "PSO_Predicted_Divorced_Population"
    ],

    marker="s",

    linewidth=2.5,

    linestyle="--",

    color="red",

    label="PSO Predicted Population"
)

plt.title(

    "Actual vs PSO Predicted Divorced Women Population",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel(
    "States / Union Territories"
)

plt.ylabel(
    "Divorced Women Population"
)

plt.xticks(

    x,

    result_df["State/UT"],

    rotation=70,

    ha="right"
)

plt.legend()

plt.grid(
    alpha=0.3
)

save_and_show(
    "result_graph.png"
)


# ================================================================
# 22. RETRAIN FINAL PSO MODEL ON COMPLETE DATASET
# ================================================================

print("\nRetraining PSO Model on Complete Dataset...")

final_pso_model = create_pipeline(
    best_params
)

final_pso_model.fit(
    X,
    y
)

model_path = output_path(
    "model.pkl"
)

joblib.dump(
    final_pso_model,
    model_path
)

print("\nPSO Model Saved:")

print(model_path)


# ================================================================
# 23. GENERATE PSO PREDICTION CSV
# ================================================================

print("\nGenerating PSO Prediction CSV...")

all_predictions = np.maximum(

    final_pso_model.predict(X),

    0
)

prediction_df = pd.DataFrame({

    "State/UT":

        df[STATE_COLUMN].values,

    "Widowed_Population":

        df[WIDOWED_COLUMN].values,

    "Actual_Divorced_Population":

        df[DIVORCED_COLUMN].values,

    "PSO_Predicted_Divorced_Population":

        all_predictions

})

prediction_df["Prediction_Difference"] = (

    prediction_df["Actual_Divorced_Population"]

    -

    prediction_df["PSO_Predicted_Divorced_Population"]

)

prediction_df["Absolute_Error"] = np.abs(
    prediction_df["Prediction_Difference"]
)

prediction_df = prediction_df.sort_values(
    by="State/UT"
)

prediction_df.to_csv(

    output_path(
        "prediction.csv"
    ),

    index=False
)

print("\nPSO Prediction CSV Saved:")

print(
    output_path("prediction.csv")
)

print("\nPrediction Preview:")

print(prediction_df.head(10))


# ================================================================
# 24. PSO PREDICTION GRAPH
# ================================================================

print("\nGenerating PSO Prediction Graph...")

plot_df = prediction_df.sort_values(

    by="PSO_Predicted_Divorced_Population",

    ascending=True
)

plt.figure(
    figsize=(16, 10)
)

y_positions = np.arange(
    len(plot_df)
)

plt.barh(

    y_positions,

    plot_df[
        "PSO_Predicted_Divorced_Population"
    ],

    color=sns.color_palette(
        "viridis",
        len(plot_df)
    ),

    edgecolor="black"
)

plt.yticks(

    y_positions,

    plot_df["State/UT"],

    fontsize=9
)

plt.xlabel(
    "PSO Predicted Divorced Women Population"
)

plt.ylabel(
    "States / Union Territories"
)

plt.title(

    "PSO State-Wise Predicted Divorced Women Population",

    fontsize=16,

    fontweight="bold"
)

plt.grid(
    axis="x",
    alpha=0.3
)

save_and_show(
    "prediction_graph.png"
)


# ================================================================
# 25. SAVE PSO MODEL JSON
# ================================================================

print("\nSaving PSO Model JSON...")

json_data = {

    "project_name":
        "AI-Based State-Wise Widowed and Divorced Women Population Analysis",

    "optimization_algorithm":
        "Particle Swarm Optimization",

    "machine_learning_model":
        "Random Forest Regressor",

    "dataset": {

        "filename":
            os.path.basename(DATASET_PATH),

        "total_records":
            int(len(df)),

        "input_feature":
            WIDOWED_COLUMN,

        "target":
            DIVORCED_COLUMN

    },

    "pso_configuration": {

        "particles":
            NUM_PARTICLES,

        "iterations":
            MAX_ITERATIONS,

        "inertia_weight":
            INERTIA_WEIGHT,

        "cognitive_coefficient":
            COGNITIVE_COEFFICIENT,

        "social_coefficient":
            SOCIAL_COEFFICIENT,

        "cv_folds":
            CV_FOLDS,

        "parameter_bounds":
            PARAMETER_BOUNDS

    },

    "best_hyperparameters":
        best_params,

    "best_cross_validation_mae":
        float(global_best_score),

    "baseline_metrics":
        baseline_metrics,

    "pso_metrics":
        pso_metrics,

    "optimization_history": {

        "global_best_cv_mae":
            global_best_history,

        "iteration_best_cv_mae":
            iteration_best_history,

        "mean_particle_cv_mae":
            mean_fitness_history

    },

    "unique_evaluations":
        int(len(fitness_cache)),

    "model_file":
        "pso_model.pkl"

}

with open(

    output_path(
        "model.json"
    ),

    "w",

    encoding="utf-8"

) as file:

    json.dump(

        json_data,

        file,

        indent=4
    )

print(
    "Saved:",
    output_path("model.json")
)


# ================================================================
# 26. SAVE PSO MODEL YAML
# ================================================================

print("\nSaving PSO Model YAML...")

yaml_data = {

    "project": {

        "name":
            "AI-Based State-Wise Widowed and Divorced Women Population Analysis",

        "optimization_algorithm":
            "Particle Swarm Optimization",

        "machine_learning_model":
            "Random Forest Regressor"

    },

    "dataset": {

        "records":
            int(len(df)),

        "input_feature":
            WIDOWED_COLUMN,

        "target":
            DIVORCED_COLUMN

    },

    "pso_configuration": {

        "particles":
            NUM_PARTICLES,

        "iterations":
            MAX_ITERATIONS,

        "inertia_weight":
            INERTIA_WEIGHT,

        "cognitive_coefficient":
            COGNITIVE_COEFFICIENT,

        "social_coefficient":
            SOCIAL_COEFFICIENT

    },

    "best_parameters":
        best_params,

    "best_cv_mae":
        float(global_best_score),

    "evaluation": {

        "baseline":
            baseline_metrics,

        "pso_optimized":
            pso_metrics

    },

    "saved_model":
        "pso_model.pkl"

}

with open(

    output_path(
        "model.yaml"
    ),

    "w",

    encoding="utf-8"

) as file:

    yaml.safe_dump(

        yaml_data,

        file,

        sort_keys=False
    )

print(
    "Saved:",
    output_path("model.yaml")
)


# ================================================================
# 27. SAVE PSO H5 FILE
# ================================================================

print("\nSaving PSO H5 File...")

# H5 stores PSO optimization results and metadata.
# The trained Random Forest is saved separately as PKL.

with h5py.File(

    output_path(
        "model.h5"
    ),

    "w"

) as h5_file:

    h5_file.attrs[
        "optimization_algorithm"
    ] = "Particle Swarm Optimization"

    h5_file.attrs[
        "machine_learning_model"
    ] = "Random Forest Regressor"

    h5_file.attrs[
        "best_cv_mae"
    ] = float(global_best_score)

    h5_file.attrs[
        "num_particles"
    ] = NUM_PARTICLES

    h5_file.attrs[
        "max_iterations"
    ] = MAX_ITERATIONS

    h5_file.create_dataset(

        "global_best_cv_mae",

        data=np.array(
            global_best_history
        )
    )

    h5_file.create_dataset(

        "iteration_best_cv_mae",

        data=np.array(
            iteration_best_history
        )
    )

    h5_file.create_dataset(

        "mean_particle_cv_mae",

        data=np.array(
            mean_fitness_history
        )
    )

    parameters_group = (
        h5_file.create_group(
            "best_parameters"
        )
    )

    for key, value in best_params.items():

        parameters_group.attrs[key] = int(
            value
        )

    metrics_group = (
        h5_file.create_group(
            "evaluation_metrics"
        )
    )

    for key, value in pso_metrics.items():

        metrics_group.attrs[key] = float(
            value
        )

print(
    "Saved:",
    output_path("model.h5")
)


# ================================================================
# 28. VERIFY GENERATED FILES
# ================================================================

print("\n" + "=" * 70)

print("VERIFYING PSO OUTPUT FILES")

print("=" * 70)

generated_files = [

    "pso_model.pkl",

    "pso_model.h5",

    "pso_model.yaml",

    "pso_model.json",

    "pso_accuracy_graph.png",

    "pso_heatmap.png",

    "pso_comparison_graph.png",

    "pso_result.csv",

    "pso_result_graph.png",

    "pso_prediction.csv",

    "pso_prediction_graph.png"

]

for filename in generated_files:

    path = os.path.join(
        BASE_DIR,
        filename
    )

    if os.path.exists(path):

        size = os.path.getsize(
            path
        ) / 1024

        print(
            f"[SUCCESS] {filename} ({size:.2f} KB)"
        )

    else:

        print(
            f"[FAILED] {filename}"
        )


# ================================================================
# 29. FINAL SUMMARY
# ================================================================

print("\n" + "=" * 70)

print("PSO PROJECT EXECUTION COMPLETED")

print("=" * 70)

print("\nOptimization Algorithm:")
print("Particle Swarm Optimization")

print("\nMachine Learning Algorithm:")
print("Random Forest Regressor")

print("\nBest Hyperparameters:")
print(best_params)

print("\nBest Cross-Validation MAE:")
print(round(global_best_score, 4))

print("\nBaseline Performance:")

for metric, value in baseline_metrics.items():

    print(
        f"{metric}: {value:.4f}"
    )

print("\nPSO Optimized Performance:")

for metric, value in pso_metrics.items():

    print(
        f"{metric}: {value:.4f}"
    )

print("\nUnique PSO Evaluations:")
print(len(fitness_cache))

print("\nAll Files Saved In:")
print(BASE_DIR)

print("\nIMPORTANT NOTES:")

print(
    "1. PSO minimizes cross-validation MAE."
)

print(
    "2. Lower MAE and RMSE indicate better regression performance."
)

print(
    "3. R2 measures regression performance, not classification accuracy."
)

print(
    "4. Test data is excluded from PSO optimization."
)

print(
    "5. Full-dataset predictions are fitted estimates, not future forecasts."
)

print(
    "6. PSO optimization does not guarantee improved test performance."
)

print("\nThank You!")
```
