```python
import os
import json
import yaml
import joblib
import warnings

import numpy as np
import pandas as pd

from sklearn.model_selection import (
    train_test_split,
    KFold,
    cross_val_score
)

from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder
)

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.impute import SimpleImputer

from sklearn.ensemble import (
    RandomForestRegressor,
    GradientBoostingRegressor
)

from sklearn.tree import DecisionTreeRegressor
from sklearn.linear_model import LinearRegression
from sklearn.svm import SVR

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Input
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")

# ============================================================
# 1. PROJECT DIRECTORY
# ============================================================

BASE_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based State-Wise Widowed and Divorced Women Population Analysis"
)

DATASET_PATH = os.path.join(
    BASE_DIR,
    "RS_Session_259_AU_3352_A_and_B.csv"
)

os.makedirs(BASE_DIR, exist_ok=True)

print("=" * 65)
print("AI-BASED WOMEN DEMOGRAPHIC POPULATION ANALYSIS")
print("=" * 65)

# ============================================================
# 2. LOAD DATASET
# ============================================================

print("\nLoading dataset...")

try:
    df = pd.read_csv(DATASET_PATH, encoding="utf-8-sig")

except UnicodeDecodeError:
    df = pd.read_csv(DATASET_PATH, encoding="latin1")

print("\nDataset loaded successfully!")

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Columns:")
print(df.columns.tolist())

print("\nFirst Five Records:")
print(df.head())

# ============================================================
# 3. DATA PREPROCESSING
# ============================================================

print("\nPerforming Data Preprocessing...")

# Remove leading/trailing spaces from column names
df.columns = df.columns.str.strip()

# Expected columns
STATE_COLUMN = "State/ UT"

WIDOWED_COLUMN = "Female of all Ages - Widowed"

DIVORCED_COLUMN = "Female of all Ages - Divorced"

required_columns = [
    STATE_COLUMN,
    WIDOWED_COLUMN,
    DIVORCED_COLUMN
]

# Check whether required columns exist
missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing columns: {missing_columns}\n"
        f"Available columns: {df.columns.tolist()}"
    )

# Remove duplicate rows
df = df.drop_duplicates()

# Clean state names
df[STATE_COLUMN] = (
    df[STATE_COLUMN]
    .astype(str)
    .str.strip()
)

# Convert numeric columns
for col in [WIDOWED_COLUMN, DIVORCED_COLUMN]:

    df[col] = pd.to_numeric(
        df[col].astype(str).str.replace(",", "", regex=False),
        errors="coerce"
    )

# Remove rows with missing target
df = df.dropna(subset=[DIVORCED_COLUMN])

# Remove invalid negative counts
df = df[df[DIVORCED_COLUMN] >= 0]

df = df.reset_index(drop=True)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDataset After Cleaning:")
print(df.shape)

if len(df) < 10:
    raise ValueError(
        "Insufficient records for reliable model training."
    )

# ============================================================
# 4. DEFINE FEATURES AND TARGET
# ============================================================

print("\nPreparing Features and Target...")

# Predict divorced population using widowed population
# and state information.

X = df[
    [
        STATE_COLUMN,
        WIDOWED_COLUMN
    ]
]

y = df[DIVORCED_COLUMN].astype(float)

print("\nInput Features:")
print(X.head())

print("\nTarget:")
print(y.head())

# ============================================================
# 5. TRAIN TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)

print("\nTraining Samples:", len(X_train))
print("Testing Samples:", len(X_test))

# ============================================================
# 6. FEATURE PREPROCESSING PIPELINE
# ============================================================

numeric_features = [WIDOWED_COLUMN]

categorical_features = [STATE_COLUMN]

numeric_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)

categorical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            numeric_transformer,
            numeric_features
        ),
        (
            "cat",
            categorical_transformer,
            categorical_features
        )
    ]
)

# ============================================================
# 7. MACHINE LEARNING MODELS
# ============================================================

models = {

    "Linear Regression": LinearRegression(),

    "Decision Tree": DecisionTreeRegressor(
        max_depth=4,
        random_state=42
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=100,
        max_depth=5,
        random_state=42
    ),

    "Gradient Boosting": GradientBoostingRegressor(
        n_estimators=100,
        learning_rate=0.05,
        max_depth=3,
        random_state=42
    ),

    "Support Vector Regression": SVR(
        kernel="rbf",
        C=100,
        epsilon=0.1
    )
}

# ============================================================
# 8. TRAIN AND EVALUATE MODELS
# ============================================================

print("\nTraining Machine Learning Models...")

results = []

trained_models = {}

for name, model in models.items():

    print("\nTraining:", name)

    pipeline = Pipeline(
        steps=[
            (
                "preprocessor",
                preprocessor
            ),
            (
                "regressor",
                model
            )
        ]
    )

    pipeline.fit(X_train, y_train)

    predictions = pipeline.predict(X_test)

    # Population cannot be negative
    predictions = np.maximum(predictions, 0)

    mae = mean_absolute_error(
        y_test,
        predictions
    )

    mse = mean_squared_error(
        y_test,
        predictions
    )

    rmse = np.sqrt(mse)

    r2 = r2_score(
        y_test,
        predictions
    )

    # Cross validation on training data
    cv = KFold(
        n_splits=5,
        shuffle=True,
        random_state=42
    )

    cv_scores = cross_val_score(
        pipeline,
        X_train,
        y_train,
        cv=cv,
        scoring="neg_mean_absolute_error"
    )

    cv_mae = -np.mean(cv_scores)

    results.append({

        "Model": name,

        "MAE": float(mae),

        "MSE": float(mse),

        "RMSE": float(rmse),

        "R2": float(r2),

        "CV_MAE": float(cv_mae)

    })

    trained_models[name] = pipeline

    print("MAE:", round(mae, 4))
    print("RMSE:", round(rmse, 4))
    print("R2 Score:", round(r2, 4))
    print("CV MAE:", round(cv_mae, 4))

# ============================================================
# 9. SELECT BEST MACHINE LEARNING MODEL
# ============================================================

results_df = pd.DataFrame(results)

# Select model using cross-validation error
results_df = results_df.sort_values(
    by="CV_MAE",
    ascending=True
)

best_model_name = results_df.iloc[0]["Model"]

best_model = trained_models[best_model_name]

print("\n" + "=" * 65)

print("BEST MACHINE LEARNING MODEL:", best_model_name)

print("=" * 65)

print(results_df.to_string(index=False))

# ============================================================
# 10. SAVE MACHINE LEARNING MODEL AS PKL
# ============================================================

pkl_path = os.path.join(
    BASE_DIR,
    "model.pkl"
)

# Refit selected model using all available data
best_model.fit(X, y)

joblib.dump(
    best_model,
    pkl_path
)

print("\nSaved:", pkl_path)

# ============================================================
# 11. PREPARE DATA FOR DEEP LEARNING
# ============================================================

print("\nPreparing Deep Learning Model...")

# Fit preprocessing on training data only
dl_preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            numeric_transformer,
            numeric_features
        ),
        (
            "cat",
            categorical_transformer,
            categorical_features
        )
    ]
)

X_train_dl = dl_preprocessor.fit_transform(X_train)

X_test_dl = dl_preprocessor.transform(X_test)

# Scale target for stable neural network training
target_scaler = StandardScaler()

y_train_dl = target_scaler.fit_transform(
    y_train.to_numpy().reshape(-1, 1)
)

y_test_dl = target_scaler.transform(
    y_test.to_numpy().reshape(-1, 1)
)

input_dimension = X_train_dl.shape[1]

# ============================================================
# 12. BUILD NEURAL NETWORK
# ============================================================

tf.random.set_seed(42)

np.random.seed(42)

dl_model = Sequential([

    Input(shape=(input_dimension,)),

    Dense(
        64,
        activation="relu"
    ),

    Dropout(0.2),

    Dense(
        32,
        activation="relu"
    ),

    Dense(
        16,
        activation="relu"
    ),

    Dense(
        1,
        activation="linear"
    )

])

dl_model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="mse",

    metrics=["mae"]

)

print("\nNeural Network Architecture:")

dl_model.summary()

# ============================================================
# 13. TRAIN DEEP LEARNING MODEL
# ============================================================

early_stopping = EarlyStopping(

    monitor="val_loss",

    patience=20,

    restore_best_weights=True

)

history = dl_model.fit(

    X_train_dl,

    y_train_dl,

    epochs=200,

    batch_size=8,

    validation_split=0.2,

    callbacks=[early_stopping],

    verbose=1

)

# ============================================================
# 14. EVALUATE DEEP LEARNING MODEL
# ============================================================

dl_predictions_scaled = dl_model.predict(
    X_test_dl,
    verbose=0
)

dl_predictions = target_scaler.inverse_transform(
    dl_predictions_scaled
).flatten()

dl_predictions = np.maximum(
    dl_predictions,
    0
)

dl_mae = mean_absolute_error(
    y_test,
    dl_predictions
)

dl_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        dl_predictions
    )
)

dl_r2 = r2_score(
    y_test,
    dl_predictions
)

print("\nDeep Learning Model Results:")

print("MAE:", dl_mae)

print("RMSE:", dl_rmse)

print("R2:", dl_r2)

# ============================================================
# 15. SAVE DEEP LEARNING MODEL AS H5
# ============================================================

h5_path = os.path.join(
    BASE_DIR,
    "model.h5"
)

dl_model.save(h5_path)

print("\nSaved:", h5_path)

# Save preprocessing objects required for predictions
joblib.dump(

    {
        "preprocessor": dl_preprocessor,
        "target_scaler": target_scaler
    },

    os.path.join(
        BASE_DIR,
        "dl_preprocessing.pkl"
    )

)

# ============================================================
# 16. CREATE YAML CONFIGURATION
# ============================================================

print("\nCreating YAML Configuration...")

yaml_data = {

    "project": {

        "name":
        "AI-Based State-Wise Widowed and Divorced Women Population Analysis",

        "type":
        "Machine Learning Regression",

        "dataset":
        os.path.basename(DATASET_PATH)

    },

    "dataset_information": {

        "total_records": int(len(df)),

        "total_features": int(X.shape[1]),

        "target_column": DIVORCED_COLUMN,

        "input_features": [
            STATE_COLUMN,
            WIDOWED_COLUMN
        ]

    },

    "training_configuration": {

        "test_size": 0.25,

        "random_state": 42,

        "cross_validation_folds": 5

    },

    "best_machine_learning_model": {

        "name": best_model_name,

        "cv_mae": float(
            results_df.iloc[0]["CV_MAE"]
        ),

        "test_mae": float(
            results_df.iloc[0]["MAE"]
        ),

        "test_rmse": float(
            results_df.iloc[0]["RMSE"]
        ),

        "test_r2": float(
            results_df.iloc[0]["R2"]
        )

    },

    "deep_learning_model": {

        "architecture": [
            64,
            32,
            16,
            1
        ],

        "optimizer": "Adam",

        "loss": "Mean Squared Error",

        "epochs": 200,

        "batch_size": 8,

        "test_mae": float(dl_mae),

        "test_rmse": float(dl_rmse),

        "test_r2": float(dl_r2)

    },

    "output_files": {

        "machine_learning_model": "model.pkl",

        "deep_learning_model": "model.h5",

        "deep_learning_preprocessing": "dl_preprocessing.pkl",

        "yaml_configuration": "model.yaml",

        "json_configuration": "model.json"

    }

}

yaml_path = os.path.join(
    BASE_DIR,
    "model.yaml"
)

with open(
    yaml_path,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        yaml_data,
        file,
        sort_keys=False,
        allow_unicode=True
    )

print("Saved:", yaml_path)

# ============================================================
# 17. CREATE JSON MODEL INFORMATION
# ============================================================

print("\nCreating JSON Model Information...")

json_data = {

    "project_name":
    "AI-Based State-Wise Widowed and Divorced Women Population Analysis",

    "dataset": {

        "filename":
        os.path.basename(DATASET_PATH),

        "records": int(len(df)),

        "features": [
            STATE_COLUMN,
            WIDOWED_COLUMN
        ],

        "target": DIVORCED_COLUMN

    },

    "machine_learning": {

        "best_model": best_model_name,

        "model_results": results_df.to_dict(
            orient="records"
        )

    },

    "deep_learning": {

        "architecture":
        json.loads(dl_model.to_json()),

        "evaluation": {

            "mae": float(dl_mae),

            "rmse": float(dl_rmse),

            "r2": float(dl_r2)

        },

        "training": {

            "epochs_completed":
            len(history.history["loss"]),

            "final_training_loss":
            float(history.history["loss"][-1]),

            "final_validation_loss":
            float(history.history["val_loss"][-1])

        }

    },

    "saved_models": {

        "machine_learning": "model.pkl",

        "deep_learning": "model.h5"

    }

}

json_path = os.path.join(
    BASE_DIR,
    "model.json"
)

with open(
    json_path,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        json_data,
        file,
        indent=4,
        ensure_ascii=False
    )

print("Saved:", json_path)

# ============================================================
# 18. VERIFY GENERATED FILES
# ============================================================

print("\n" + "=" * 65)

print("VERIFYING GENERATED MODEL FILES")

print("=" * 65)

output_files = [

    "model.h5",

    "model.pkl",

    "model.yaml",

    "model.json",

    "dl_preprocessing.pkl"

]

for filename in output_files:

    file_path = os.path.join(
        BASE_DIR,
        filename
    )

    if os.path.exists(file_path):

        file_size = os.path.getsize(
            file_path
        ) / 1024

        print(
            f"[SUCCESS] {filename} "
            f"({file_size:.2f} KB)"
        )

    else:

        print(
            f"[FAILED] {filename}"
        )

# ============================================================
# 19. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 65)

print("PROJECT EXECUTION COMPLETED")

print("=" * 65)

print("\nBest ML Model:", best_model_name)

print(
    "Best ML Cross-Validation MAE:",
    round(
        float(results_df.iloc[0]["CV_MAE"]),
        4
    )
)

print("\nDeep Learning MAE:", round(dl_mae, 4))

print("Deep Learning RMSE:", round(dl_rmse, 4))

print("Deep Learning R2:", round(dl_r2, 4))

print("\nAll models saved in:")

print(BASE_DIR)

print("\nNOTE:")
print(
    "This is an exploratory demographic regression project. "
    "The small dataset does not support reliable future "
    "population forecasting."
)
```
